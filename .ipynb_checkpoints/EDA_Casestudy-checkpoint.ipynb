```python
# Data visualization --> mathplotlib ,seaborn
import matplotlib.pyplot as plt 
```

```python
plt.hist(data['Age'],bins=15,color='orange') #histogram is used for numerical data and bins is used for number of bars count 
plt.xlabel("Age")
plt.xlabel("Count")
plt.title("Age Distributrion")
plt.show() #to remove the data above the graph like only graph noo data
```
